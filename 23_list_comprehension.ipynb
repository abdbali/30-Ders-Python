**Günün Hedefleri:**
- Klasik `for` döngüsü ile List Comprehension arasındaki farkı görmek.

- Oluşturulan listeye `if` şartları (Örn: Sadece çift sayıları ekle) yerleştirmek.

## 1. Uzun Yol vs. Kısa Yol (List Comprehension)

Amacımız: 0'dan 5'e kadar olan sayılardan bir liste oluşturmak.

```python
# --- UZUN YOL (Eski Yöntem) ---
eski_liste = []
for i in range(5):
    eski_liste.append(i)

print("Eski Liste:", eski_liste)

# --- KISA YOL (List Comprehension) ---
# Formül: [ne_eklenecek for dongu_degiskeni in liste_veya_range]
yeni_liste = [i for i in range(5)]

print("Yeni Liste:", yeni_liste)
```

## 2. Eklemeden Önce İşlem Yapmak

Sadece `i` eklemek zorunda değiliz. Listenin içine atacağımız değere baştaki kısımda matematiksel işlemler uygulayabiliriz. 
Örneğin sayıların kendisini değil, karesini (`i * i` veya `i**2`) listeye atabiliriz.

```python
# 1'den 5'e kadar (5 dahil) sayıların karelerinden oluşan bir liste:
kareler_listesi = [sayi * sayi for sayi in range(1, 6)]

print("Kareler Listesi:", kareler_listesi)

# Stringler üzerinde işlem yapmak:
isimler = ["ali", "ayşe", "veli"]
buyuk_isimler = [isim.upper() for isim in isimler]
print("Büyük Harfli İsimler:", buyuk_isimler)
```

## 3. Şartlı Kısayol (if ile Filtreleme)

List Comprehension'ın en güçlü yanı `if` ile filtreleme yapabilmesidir. Formülün en sağına bir `if` şartı eklersen, sadece o şartı sağlayanlar listeye dahil edilir.

Formül: `[ne_eklenecek for dongu in kaynak if şart]`

```python
# 0'dan 10'a kadar olan sayılardan SADECE ÇİFT (2'ye bölünen) olanları almak:
cift_sayilar = [x for x in range(11) if x % 2 == 0]
print("Çift Sayılar:", cift_sayilar)

# Notu 50'den büyük olan (Geçen) öğrencileri yeni listeye aktarmak:
ogrenci_notlari = [45, 80, 20, 95, 60, 30]
gecen_notlar = [notu for notu in ogrenci_notlari if notu >= 50]
print("Dersi Geçen Notlar:", gecen_notlar)
```
