## 1. Local (Yerel) Değişkenler

Bir fonksiyonun KENDİ İÇİNDE (def bloğunun girintisi altında) oluşturulan değişkenlere **Local (Yerel) Değişkenler** denir.
Bu değişkenler fonksiyon çalıştığı an doğarlar ve fonksiyonun `return` (veya bitiş) satırında ÖLÜRLER. Dışarıdan onlara ASLA ulaşılamaz.

```python
def iksir_yap():
    iksir_gucu = 50  # Bu bir Local (Yerel) değişkendir
    print(f"İksir üretildi, gücü: {iksir_gucu}")

iksir_yap()

# ŞİMDİ HATA ZAMANI! (Kodu çalıştırıp hatayı incele)
# Dışarıdan, fonksiyonun içindeki gizli bir değişkene ulaşmaya çalışıyoruz:
# print("Dışarıdaki iksir gücü:", iksir_gucu) 
# NameError: name 'iksir_gucu' is not defined hatası alırsın!
```

## 2. Global (Evrensel) Değişkenler

Sayfanın en başında, hiçbir fonksiyonun veya bloğun İÇİNDE OLMADAN dümdüz tanımladığımız değişkenlere **Global Değişkenler** denir.

Global değişkenler sokağın ortasındaki tabela gibidir. Fonksiyonlar (evler) pencerelerinden dışarı bakıp o tabelayı GÖREBİLİRLER ve OKUYABİLİRLER.

```python
oyun_seviyesi = 5  # Bu bir Global değişkendir. Herkes görebilir.

def seviye_sorgula():
    # Fonksiyon içinden dışarıdaki (global) değişkene sadece BAKABİLİYORUZ.
    print(f"Oyuncunun şu anki seviyesi: {oyun_seviyesi}")

seviye_sorgula()
```

## 3. Global Değişkeni Değiştirmek ve "global" Komutu

Fonksiyonlar dışarıdaki (Global) değişkene sadece BAKABİLİR demiştik. Peki ya oyun seviyesi artarsa ve fonksiyonun içinden bu dışarıdaki `oyun_seviyesi`ni değiştirmek istersek?

Eğer direkt `oyun_seviyesi = 6` yazarsan, Python yeni bir tane Local (yerel) kopya oluşturur, dışarıdakini ellemez. 
Dışarıdakini değiştirebilmek için Python'a özel bir mesaj vermeliyiz: `global oyun_seviyesi` (Ben dışarıdaki orijinaline müdahale edeceğim, yenisini oluşturma!)

```python
kasa_parasi = 100

def para_kazan():
    # Dışarıdaki asıl kasayı güncelleyeceğimizi belirtiyoruz
    global kasa_parasi  
    kasa_parasi += 50
    print("Kasaya para eklendi!")

para_kazan()
print("Dışarıdan Baktığımızda Güncel Kasa:", kasa_parasi)

## DİKKAT: Büyük ve profesyonel yazılımlarda 'global' kullanmaktan OLABİLDİĞİNCE KAÇINILIR.
## Bunun yerine veriler fonksiyona parametre olarak sokulur ve 'return' ile dışarı çıkartılıp güncellenir.
```
