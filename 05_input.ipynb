## 1. input() Nedir?

`print()` fonksiyonu bilgisayarın bizimle konuşmasını sağlarken, `input()` fonksiyonu bilgisayarın bizden cevap beklemesini sağlar. `input` İngilizcede 'girdi' veya 'giriş' demektir.

Program `input()` satırına geldiğinde, kullanıcının klavyeden bir şeyler yazıp 'Enter' tuşuna basmasını bekler. Enter'a basılana kadar kodun alt satırlarına inmez, programı orada duraklatır.

```python
# Bilgisayar soruyu sorar ve kullanıcının cevabını 'isim' isimli kutuya (değişkene) koyar.
isim = input("Lütfen isminizi girin: ")

# Kullanıcı cevabı girip Enter'a bastıktan sonra aşağıdaki kod çalışır.
print(f"Merhaba {isim}! Benimle konuştuğun için çok sevindim.")
```

## 2. DİKKAT: input() Her Zaman String'dir!

Kullanıcıya yaşını sorduğumuzda ve kullanıcı ekrana 15 yazdığında, Python bunu matematiksel olan 15 sayısı olarak (Integer) görmez. Bunu harflerden oluşan "15" (String) metni olarak görür.

Yani tırnak içindeki "15" ile, matematiğe dahil olan 15 sayıları Python için birbirinden tamamen farklı şeylerdir.

```python
yas = input("Yaşınızı girin: ")

# type() fonksiyonu ile gelen verinin türüne bakalım
print("Girdiğiniz yaşın türü:", type(yas))

# Eğer bu yaşın üstüne sayısal bir işlem yapmaya kalkarsak hata alırız.
# print(yas + 5)  # TypeError hatası verir çünkü string ile integer toplanamaz.
```

## 3. Tür Dönüşümü (Type Casting)

Peki matematikte kullanmak üzere klavyeden bir sayı almak istiyorsak ne yapacağız? Gelen String (metin) verisini zorla Integer (sayı) verisine dönüştüreceğiz! Buna Type Casting (Tür Dönüşümü) denir.

- `int(veri)` : Veriyi tam sayıya çevirir.
- `float(veri)` : Veriyi ondalıklı sayıya çevirir.
- `str(veri)` : Veriyi metne çevirir.

```python
yas_metin = input("Lütfen yaşınızı girin: ")

# String olan yaşı Integer'a çeviriyoruz
yas_sayi = int(yas_metin)

# Artık güvenle matematiksel işlem yapabiliriz
gelecekteki_yas = yas_sayi + 10
print(f"10 yıl sonraki yaşınız: {gelecekteki_yas}")
```

## 4. input() ve Dönüşümü Aynı Satırda Yapmak (Kısa Yol)

Yazılımcılar uzun kod yazmayı sevmezler. Input aldığımız satırda anında onu `int()` fonksiyonu ile sararsak, daha veriyi alır almaz sayıya çevirmiş oluruz.

```python
# input fonksiyonunu doğrudan int() parantezlerinin içine yazıyoruz.
dogum_yili = int(input("Hangi yıl doğdunuz? (Örn: 2010): "))

yas = 2024 - dogum_yili
print(f"Siz şu anda {yas} yaşındasınız.")
```
