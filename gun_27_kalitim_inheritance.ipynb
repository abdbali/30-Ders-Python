- `super().__init__()` fonksiyonu ile babanın genlerini alt sınıfa kopyalamak.

## 1. Üst Sınıf (Parent) Oluşturmak

Önce bütün araçların ortak özelliklerini barındıran genel bir `Tasit` (Araç) sınıfı tasarlayalım.

```python
class Tasit:  # Ana sınıfımız (Parent)
    def __init__(self, marka, yil):
        self.marka = marka
        self.yil = yil
        
    def korna_cal(self):
        print("Düüüüt!")

print("Ana sınıf (Taşıt) hazırlandı!")
```

## 2. Miras Almak ve super() Kullanımı

Şimdi bu genel taşıttan bir `Araba` (Child) sınıfı türeteceğiz. Araba da bir taşıttır, yani markası ve yılı vardır. 
Parantez içine `(Tasit)` yazarak, Taşıt'ın tüm kodlarını kopyalıyoruz. 
Sonra `super()` (Üst Sınıfı çağır) komutuyla o özellikleri kendimize mal ediyoruz.

```python
# Araba sınıfı, Tasit sınıfından MIRAS ALIYOR
class Araba(Tasit):
    def __init__(self, marka, yil, kapi_sayisi):
        # super() diyerek Tasit'in __init__ metodunu çalıştırıyoruz. 
        # Yani marka ve yılı tekrar self.marka = marka diye yazmamıza gerek kalmıyor!
        super().__init__(marka, yil)
        
        # Kendisine has yepyeni bir özellik
        self.kapi_sayisi = kapi_sayisi
        
    # Kendisine has özel bir metot
    def drift_yap(self):
        print(f"{self.marka} virajı drift yaparak dönüyor! Sckrrr!")

# Şimdi bir Araba üretelim
spor_araba = Araba("Ferrari", 2023, 2)

print("Arabanın Yılı:", spor_araba.yil) # Yılı Tasit'tan aldı!
spor_araba.korna_cal() # Korbayı Tasit'tan aldı!
spor_araba.drift_yap() # Drifiti KENDİSİ yaptı!
```

## 3. Override (Üst Sınıfın Özelliğini Ezmek)

Eğer `Araba` sınıfının içine de bir `korna_cal` metodu yazarsan, `Araba` üretildiğinde Taşıtın değil, KENDİ yazdığı kornayı kullanır. Buna "Override" (Üstüne yazmak/Ezmek) denir.

```python
class Bisiklet(Tasit):
    def __init__(self, marka, yil):
        super().__init__(marka, yil)
        
    # Taşıttaki Düüüt kornasını eziyoruz (Override)
    def korna_cal(self):
        print("Zırrr Zırrr! (Bisiklet zili)")

bisikletim = Bisiklet("Salcano", 2020)
bisikletim.korna_cal() # Artık Düüüt demez, Zırrr der!
```
