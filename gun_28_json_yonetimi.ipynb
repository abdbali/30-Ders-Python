**Günün Hedefleri:**
- JSON mantığını anlamak.
- `import json` ile JSON kütüphanesini projeye eklemek.
- `json.dump()` ile Python sözlüklerini/listelerini `.json` dosyasına güvenle kaydetmek.
- `json.load()` ile dosyadan veriyi bozmadan Python Sözlüğüne geri çevirmek.

## 1. JSON Kaydetmek (json.dump)

Elimizde çok detaylı bir oyuncu sözlüğü (Dict) var. Bunu `"w"` (yazma) moduyla bir `.json` dosyası açarak `json.dump(veri, dosya)` komutu ile içine gömeceğiz.

```python
import json

# Elimizdeki Python Sözlüğü (Dict)
oyuncu_verisi = {
    "isim": "Kratos",
    "seviye": 50,
    "esyalar": ["Balta", "Kalkan", "İksir"], # Liste bile var!
    "oldu_mu": False
}

# 'oyuncu_kayit.json' adında dosya açıp, içine sözlüğü DUMP (Boşalt/Yaz) ediyoruz
with open("oyuncu_kayit.json", "w", encoding="utf-8") as dosya:
    # indent=4 parametresi JSON dosyasının içini alt alta, okunaklı, girintili yapar.
    json.dump(oyuncu_verisi, dosya, indent=4)

print("Veri başarıyla JSON olarak kaydedildi!")
```

## 2. JSON Dosyasını Geri Okumak (json.load)

Oyunu ertesi gün açtık. Hard diskte yatan `.json` dosyasındaki yazıları, programımızın içine canlı bir Sözlük (Dict) olarak geri yüklemeliyiz.
Bunu `.read()` ile değil, `json.load()` ile yaparız.

```python
import json

# 'r' kipiyle okuma amaçlı açıyoruz
with open("oyuncu_kayit.json", "r", encoding="utf-8") as dosya:
    # Dosyanın içindekini YÜKLE (Load) ve bir değişkene at
    geri_gelen_veri = json.load(dosya)

print("Dosyadan Gelen Veri:", geri_gelen_veri)
print("Verinin Türü (Type):", type(geri_gelen_veri))

# Tam teşekküllü bir Sözlük olduğu için direkt anahtarını çağırabiliriz:
print("\nHoş geldin", geri_gelen_veri["isim"])
print("Karakterinin 1. Eşyası:", geri_gelen_veri["esyalar"][0])
```

## 3. Neden txt Değil de JSON?

Eğer aynı şeyi `txt` dosyasına yazıp `.read()` ile geri okusaydık, Python onu bir sözlük olarak değil, `"{'isim': 'Kratos'}"` şeklinde devasa ve manasız bir METİN (String) olarak algılayacaktı. `.esyalar` veya `['isim']` diyerek içindeki verilere ulaşamayacaktık. 
JSON veri yapılarını (liste, bool, dict) %100 korur!
