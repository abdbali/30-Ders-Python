## 1. Algoritma Nedir?

Algoritma, bir sorunu çözmek için atılması gereken adımların sırasıdır. 
En klasik örnek: Çay Demlemek.
1. Çaydanlığa su koy.
2. Ocağı yak ve suyu kaynat.
3. Su kaynadı mı? (Eğer Hayır ise, kaynamasını bekle - Döngü/while).
4. Su kaynadıysa demliğe çay at ve suyu üzerine dök (Eğer Evet ise - if/break).
5. 10 dakika bekle (time.sleep).
6. Çayı servis et (return).

İşte kod yazmadan önce bu düz metin (Sözde Kod / Pseudocode) taslağını oluşturmalısın.

## 2. Modüler Düşünce (Parçala ve Yönet)

Diyelim ki bir Okul Otomasyon Sistemi yapıyorsun. Tüm kodları tek bir sayfaya, tek bir `while True` döngüsünün içine yazarsan o kod ileride bir makarna ağına (Spaghetti Code) dönüşür ve çöker.

Bunun yerine sistemi modüllere (Fonksiyonlara ve Sınıflara) bölmeliyiz:
- **Modül 1 (Veri):** Ogrenci adında bir Class oluştur. (Özellikleri tutar)
- **Modül 2 (Kayıt):** ogrenci_kaydet() diye bir fonksiyon yaz (Sadece JSON'a kaydeder)
- **Modül 3 (Görüntü):** ekrana_bas() diye bir fonksiyon yaz (Sadece print yapar)
- **Modül 4 (Ana Menü):** while True döngüsü kur (Sadece yukardaki fonksiyonları çağırır)

İşte buna parçala ve yönet diyoruz. Biri bozulursa, sadece o modülü tamir edersin.

## 3. Hata Ayıklama (Debugging) Taktikleri

Kodun çalışmıyor, nedenini bulamıyorsun. Neler yapabilirsin?
1. **Bollukla Print Kullan:** Hangi fonksiyonun veya `if` bloğunun çalışmadığını anlamak için aralara `print("Buraya kadar geldi!")` veya `print(x)` yazarak değişkenlerin anlık değerlerini kontrol et.
2. **Ördek Muhabbeti (Rubber Duck Debugging):** Kodu sanki karşında hiç yazılım bilmeyen biri (veya plastik bir ördek) varmış gibi satır satır sesli oku ve anlat. %90 ihtimalle kendi hatanı sesli anlatırken bulacaksın.
3. **Hata Kodunu Okumak:** Kırmızı çıkan yazıda `Line 45` diyorsa, sorun gerçekten 45. satırda veya ondan bir önceki satırdadır (örn: parantez kapanmamıştır).
