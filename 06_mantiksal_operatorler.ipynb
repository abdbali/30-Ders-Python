## 1. Boolean (bool) Veri Tipi

Şu ana kadar String, Integer ve Float veri tiplerini öğrendik. Şimdi ise 'Boolean' ile tanışıyoruz. Boolean veri tipi, içinde sadece ve sadece iki farklı değer barındırabilir:

1. **True** (Doğru / Evet / 1)
2. **False** (Yanlış / Hayır / 0)

Not: Python'da `True` ve `False` kelimelerinin baş harfleri mutlaka büyük yazılmalıdır. Aksi takdirde Python bunları normal bir değişken zanneder ve hata verir.

```python
oyun_bitti_mi = False
oyuncu_hayatta_mi = True

print("Oyun bitti mi?", oyun_bitti_mi)
print("Bu verinin türü:", type(oyuncu_hayatta_mi))
```

## 2. Karşılaştırma Operatörleri

Karşılaştırma operatörleri bilgisayara bir soru sorar. Bilgisayar da bu soruyu hesaplar ve bize cevap olarak sadece `True` veya `False` döndürür.

- `==` : Eşit mi? (Dikkat: Tek eşittir atama yapar, iki eşittir karşılaştırma yapar!)
- `!=` : Eşit değil mi?
- `>`  : Büyük mü?
- `<`  : Küçük mü?
- `>=` : Büyük veya eşit mi?
- `<=` : Küçük veya eşit mi?

```python
oyuncu_seviyesi = 5
gerekli_seviye = 10

print("Oyuncu, gerekli seviyeye ulaştı mı? (5 >= 10):", oyuncu_seviyesi >= gerekli_seviye)

puan = 100
print("Kazanmak için gereken puan alındı mı? (100 == 100):", puan == 100)

can = 0
print("Oyuncu öldü mü? (can == 0):", can == 0)
```

## 3. Stringlerde (Metinlerde) Karşılaştırma

Sadece sayıları değil, metinleri de birbiriyle karşılaştırabiliriz. Bu, genellikle parola sistemlerinde veya kullanıcı adı kontrollerinde kullanılır. Python için büyük ve küçük harfler tamamen farklı karakterlerdir (Büyük küçük harf duyarlılığı - Case Sensitivity).

```python
sifre_veritabani = "GizliSifre123"
kullanicinin_girdigi = "gizlisifre123"

print("Şifreler tam olarak eşleşiyor mu?", sifre_veritabani == kullanicinin_girdigi)

dogru_isim = "Ahmet"
print("İsimler farklı mı?", dogru_isim != "Mehmet")
```

## 4. Mantıksal Operatörler (and, or, not)

Gerçek hayatta kararlar her zaman tek bir şarta bağlı olmaz. 'Ehliyet almak için hem 18 yaşından büyük OLMALISIN, VE hem de sınavı geçmiş OLMALISIN' gibi.

1. **`and` (VE):** Her iki şart da True ise sonuç True olur. Şartlardan biri bile False ise sonuç False'dur.
2. **`or` (VEYA):** Şartlardan sadece birinin True olması yeterlidir. İkisi birden False ise sonuç False olur.
3. **`not` (DEĞİL):** Çıkan sonucu tam tersine çevirir. True ise False, False ise True yapar.

```python
yas_uygun = True
sinavi_gecti = False

print("Ehliyet alabilir mi? (and):", yas_uygun and sinavi_gecti)

kartta_para_var = False
nakit_para_var = True

print("Alışveriş yapabilir mi? (or):", kartta_para_var or nakit_para_var)

kapi_kilitli_mi = True
print("Kapı açık mı? (not):", not kapi_kilitli_mi)
```

## 5. Operatörleri Birlikte Kullanmak (Karmaşık Karşılaştırmalar)

Karşılaştırma operatörleri ve mantıksal operatörleri aynı satırda birleştirebiliriz. Tıpkı matematikteki gibi işlemleri parantez içine almak okunabilirliği artırır.

```python
kullanici_yasi = 25
vip_uyesi_mi = False

# VIP odasına girmek için yaşın 18'den büyük olması VE (VIP üye olması VEYA yanında davetiye olması) gerekir.
# VIP değil ve davetiyesi yok, ama yaşı tutuyor. Bakalım girebilecek mi?

davetiye_var_mi = False

girebilir_mi = (kullanici_yasi >= 18) and (vip_uyesi_mi or davetiye_var_mi)
print("Kullanıcı VIP odaya girebilir mi?", girebilir_mi)
```
