## 1. Sınıf (Class) ve Nesne (Object) Nedir?

Bunu bir **Kurabiye Kalıbı** olarak düşün.
- **Sınıf (Class):** Kurabiye kalıbının kendisidir (Taslak). Şeklinin yıldız olacağını belirler ama yenmez. Kodlarda `class` kelimesiyle yazılır.
- **Nesne (Object):** O kalıbı kullanarak hamurdan kestiğin ve fırından çıkardığın gerçek kurabiyelerdir. 1 kalıpla 1000 tane farklı kurabiye (nesne) üretebilirsin!

Aşağıda en basit haliyle içi boş bir taslak (Class) oluşturalım.

```python
# class isimleri her zaman BÜYÜK HARFLE başlar (Yazılımcı kuralı)
class Kopek:
    # Şimdilik içi boş bir kalıp yapalım (pass kelimesi 'boş geç' demektir)
    pass

# Kalıbı kullanarak 2 farklı canlı (Nesne) köpek üretelim
kopek1 = Kopek()
kopek2 = Kopek()

print("Köpek 1'in Tipi:", type(kopek1)) # Tipi artık string veya list değil, kendi ürettiğimiz Kopek!
print("Köpek 1 ve 2 aynı şeyler mi?:", kopek1 == kopek2) # Hayır, ikisi farklı canlar.
```

## 2. Nesnelere Özellik (Attribute) Vermek

Fırından çıkardığımız kurabiyelere sonradan çikolata veya krema sürebiliriz. 
Ürettiğimiz nesnelere de `nesne.ozelliki_ismi = deger` diyerek tıpkı bir sözlük gibi bilgiler (özellikler) yükleyebiliriz.

```python
# kopek1 nesnemize özellikler verelim
kopek1.isim = "Karabaş"
kopek1.cins = "Kangal"
kopek1.yas = 3

# kopek2 nesnemize de farklı özellikler verelim
kopek2.isim = "Pamuk"
kopek2.yas = 1

print(f"1. Köpeğin adı: {kopek1.isim}, Cinsi: {kopek1.cins}")
print(f"2. Köpeğin adı: {kopek2.isim}, Yaşı: {kopek2.yas}")

# Kötü Haber: Özellikleri böyle sonradan tek tek eklemek uzun ve hataya açıktır. 
# Peki bunu kalıbı (class) üretirken otomatik yapsak olmaz mı?
```

## 3. İlk Otomatik Kurulum (The __init__ method)

Bir nesneyi yaratırken (kurabiye basarken) ona en başından zorunlu özelliklerini (isim, yaş vb.) vermek için Sınıf'ın içine özel bir fonksiyon (`def`) yazarız. 
Bu fonksiyonun adı her zaman `__init__` olmak zorundadır (Sağında ve solunda ikişer alt çizgi vardır). İngilizce 'Initialize' (Başlatmak/Kurmak) kelimesinden gelir.

Buradaki sihirli kelimemiz ise `self` (kendisi) kelimesidir. Nesnenin kendi kendisine özellik eklemesini sağlar.

```python
class Araba:
    # Kalıptan her yeni araba çıktığında bu fonksiyon OTOMATİK ÇALIŞIR!
    # 'self' üretilen arabanın ta kendisidir (ruhu). Diğerleri de bizim gönderdiğimiz malzemelerdir.
    def __init__(self, gelen_marka, gelen_renk):
        # Kendisine bu malzemeleri kalıcı özellik olarak ekliyor
        self.marka = gelen_marka
        self.renk = gelen_renk
        self.tekerlek_sayisi = 4 # Her arabada varsayılan olarak sabit 4 tekerlek olsun

# Arabayı yaratırken artık içine mecburen marka ve renk parametrelerini veriyoruz (self otomatik gider, biz yollamayız)
araba1 = Araba("Ford", "Kırmızı")
araba2 = Araba("BMW", "Mavi")

# Tekerlek sayısını biz vermedik ama __init__ içinde otomatik atandı!
print(f"{araba1.marka} markalı araba {araba1.renk} renklidir ve {araba1.tekerlek_sayisi} tekeri vardır.")
print(f"{araba2.marka} markalı araba {araba2.renk} renklidir.")
```
