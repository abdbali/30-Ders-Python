## 1. Listeye Yeni Eleman Eklemek

Bir listenin en sonuna (trenin en arkasına) yeni bir vagon eklemek istiyorsak `.append()` metodunu kullanırız. 
Eğer en sona değil de, istediğimiz belli bir sıraya eklemek istiyorsak `.insert(indeks_no, deger)` metodunu kullanırız.

```python
oyun_sepeti = ["Minecraft", "Roblox"]
print("Başlangıç:", oyun_sepeti)

# En sona yeni oyun ekleyelim
oyun_sepeti.append("Valorant")
print("Append sonrası:", oyun_sepeti)

# En başa (0. indekse) yeni oyun ekleyelim, diğerleri bir sağa kaysın
oyun_sepeti.insert(0, "Fifa")
print("Insert sonrası:", oyun_sepeti)
```

## 2. Listeden Eleman Silmek

Listeden eleman silmenin iki farklı yolu vardır:
1. İsmini (değerini) biliyorsan: `.remove("İsim")` kullanırsın. Eğer o isimden birden fazla varsa sadece İLK bulduğunu siler.
2. İndeks numarasını (sırasını) biliyorsan: `.pop(indeks)` kullanırsın. İçine numara yazmazsan otomatik olarak EN SONDAKİ elemanı siler.

```python
market_listesi = ["Ekmek", "Süt", "Yumurta", "Süt", "Peynir"]

# "Süt" ismindeki İLK elemanı silelim
market_listesi.remove("Süt")
print("Remove sonrası:", market_listesi)

# Sondaki elemanı silelim (Peynir)
market_listesi.pop()
print("Pop (boş) sonrası:", market_listesi)

# 1. indeksteki elemanı silelim (Yumurta)
market_listesi.pop(1)
print("Pop(1) sonrası:", market_listesi)

# Tüm listeyi tamamen boşaltmak istersek:
market_listesi.clear()
print("Clear sonrası:", market_listesi)
```

## 3. Listeyi Sıralamak ve Ters Çevirmek

Bir sayı listesini küçükten büyüğe veya bir isim listesini alfabetik olarak (A-Z) sıralamak için `.sort()` kullanılır.
Listeyi mevcut halinin tam tersine (sondan başa doğru) çevirmek için ise `.reverse()` kullanılır.

```python
notlar = [45, 90, 15, 100, 60]
isimler = ["Zehra", "Ali", "Buse"]

# Sayıları küçükten büyüğe sıralama
notlar.sort()
print("Sıralı Notlar:", notlar)

# İsimleri alfabetik sıralama
isimler.sort()
print("Sıralı İsimler:", isimler)

# Listeyi tersine çevirme (Dikkat: Sıralamaz, sadece mevcut dizilimi tersine aynalar!)
isimler.reverse()
print("Ters Çevrilmiş:", isimler)
```

## 4. Uzunluk Ölçmek ve Eleman Saymak

`len()` fonksiyonunu kelimelerin uzunluğunu bulurken öğrenmiştik. Listelerdeki vagon (eleman) sayısını bulmak için de aynısını kullanırız.
`.count()` metodu ise, listede Belli bir elemandan kaç tane geçtiğini sayar.

```python
anket = ["Evet", "Hayır", "Evet", "Evet", "Hayır"]

print("Toplam cevap sayısı (Vagon sayısı):", len(anket))

evet_sayisi = anket.count("Evet")
print(f"Ankette toplam {evet_sayisi} kişi Evet dedi.")
```
