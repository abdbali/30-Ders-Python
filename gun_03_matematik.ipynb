## 1. Temel Dört İşlem

Python'da toplama, çıkarma, çarpma ve bölme işlemleri günlük hayatta kullandığımız sembollere çok benzer. Sadece çarpma için 'x' yerine yıldız (*), bölme için ise '÷' yerine eğik çizgi (/) kullanırız.

```python
a = 20
b = 5

print("Toplama İşlemi (a + b):", a + b)
print("Çıkarma İşlemi (a - b):", a - b)
print("Çarpma İşlemi (a * b):", a * b)
print("Bölme İşlemi (a / b):", a / b)  # Dikkat: Bölme işlemi her zaman ondalıklı sayı (float) üretir!
```

## 2. Gelişmiş Operatörler

Programlamada sadece 4 işlem yetmez. Özellikle oyunlarda veya zaman hesaplamalarında çok işimize yarayan 3 özel operatörümüz daha vardır:

1. **Tam Sayı Bölmesi (//):** Bölme işlemini yapar ama küsuratı çöpe atar. Sadece tam sayıyı verir.
2. **Üs Alma (**):** Bir sayının kuvvetini almak için kullanılır. (Örneğin 2'nin 3. kuvveti = 2 * 2 * 2)
3. **Modülüs veya Mod Alma (%):** Bir bölme işleminden Kalan sayıyı verir. Programlamadaki en önemli operatörlerden biridir!

```python
x = 10
y = 3

print("Normal Bölme (10 / 3):", x / y)       # Çıktı: 3.3333333333333335
print("Tam Sayı Bölmesi (10 // 3):", x // y) # Çıktı: 3 (Küsurat silindi)

print("Üs Alma (10 ** 3):", x ** y)          # Çıktı: 1000 (10x10x10)

# Modülüs: 10'u 3'e bölersek bölüm 3, kalan ise 1'dir. Bize o kalan (1) sayısını verir.
print("Mod Alma (10 % 3):", x % y)           # Çıktı: 1
```

## 3. Modülüs (%) Neden Çok Önemlidir?

İlk bakışta 'Kalan sayıyı bulmak ne işime yarayacak?' diyebilirsin. Ancak modülüs programlamanın gizli kahramanıdır. 

En çok kullanıldığı yer: Bir sayının tek mi yoksa çift mi olduğunu bulmak!
Eğer bir sayının 2'ye bölümünden kalan 0 ise (sayi % 2), o sayı çifttir. Eğer kalan 1 ise, o sayı tektir.

```python
cift_sayi = 14
tek_sayi = 21

print("14 sayısının 2'ye bölümünden kalan:", cift_sayi % 2)
print("21 sayısının 2'ye bölümünden kalan:", tek_sayi % 2)
```

## 4. İşlem Önceliği

Matematik derslerindeki kuralları hatırlıyor musun? Çarpma işlemi toplamadan önce yapılır. Python da matematikten çok iyi anlar ve bu kurallara harfiyen uyar.

İşlem Sırası (Yukarıdan Aşağıya):
1. Parantez İçi ()
2. Üs Alma **
3. Çarpma *, Bölme /, Modülüs %, Tam Sayı Bölmesi //
4. Toplama +, Çıkarma -

```python
# Burada önce 3 * 2 yapılır (6), sonra 5 eklenir = 11
hesap_1 = 5 + 3 * 2
print("Parantezsiz işlem:", hesap_1)

# Eğer önce toplamanın yapılmasını istiyorsak, parantez kullanmalıyız!
# Burada önce (5 + 3) yapılır (8), sonra 2 ile çarpılır = 16
hesap_2 = (5 + 3) * 2
print("Parantezli işlem:", hesap_2)
```

## 5. Kısa Yoldan Atama Operatörleri

Oyunlarda karakterin puanını artırmak için genellikle 'puan = puan + 10' yazarız. Python bunu daha kısa yazmamız için bize özel araçlar sunar.

- `+=` : Topla ve ata
- `-=` : Çıkar ve ata
- `*=` : Çarp ve ata
- `/=` : Böl ve ata

```python
can = 100
print("Başlangıç canı:", can)

# Karakter hasar aldı
can -= 20   # Bu işlem 'can = can - 20' ile tamamen aynıdır.
print("Hasar alındıktan sonra can:", can)

# Karakter iksir içti
can += 50   # Bu işlem 'can = can + 50' ile aynıdır.
print("İksir içtikten sonra can:", can)
```
