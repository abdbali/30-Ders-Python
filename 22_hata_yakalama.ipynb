## 2. Kalkanımızı Kuruyoruz: try ve except

- **`try` (Dene):** "Şu kodları çalıştırmayı bir dene, eğer hata çıkarsa hiç panik yapma, programı durdurma, hemen aşağıdaki `except` bloğuna atla" demektir.
- **`except` (Hata Durumunda):** Hata olursa ekranda ne gösterileceğini yazdığımız yerdir.

```python
try:
    yas = int(input("Lütfen yaşınızı RAKAMLA girin: "))
    print(f"Bir yıl sonra {yas + 1} yaşında olacaksın.")

except:
    # Kullanıcı harf bile girse program ÇÖKMEYECEK, burası çalışacak.
    print("HATA! Lütfen kelime kullanmayın, sadece sayı (örn: 15) girin.")
    
print("Eğer bu yazıyı görüyorsan, program hata alsa bile çökmeyip hayatta kalmış demektir!")
```

## 3. Hata Türüne Göre Farklı Önlemler Almak

Sadece genel bir 'Hata var' demek bazen yetmez. Hatanın türüne göre kullanıcıya farklı cevaplar verebiliriz. En meşhur iki hata:
1. `ValueError`: Verinin türü yanlışsa (sayı yerine harf gibi).
2. `ZeroDivisionError`: Matematikte hiçbir sayı sıfıra bölünemez! Böldürmeye çalışırsan bu hatayı alırsın.

```python
try:
    sayi1 = int(input("Birinci sayıyı gir: "))
    sayi2 = int(input("Bölen sayıyı gir (Dikkat sıfır girme): "))
    
    sonuc = sayi1 / sayi2
    print("Bölme sonucu:", sonuc)

except ValueError:
    print("Uyarı: Lütfen harf değil, rakam giriniz!")
    
except ZeroDivisionError:
    print("Matematiksel Hata: Hiçbir sayı 0'a bölünemez, evren çöker!")

except Exception as e:
    # Beklenmeyen başka bir hata çıkarsa, 'e' değişkeni ile hatanın ne olduğunu ekrana yazdırırız.
    print("Bilinmeyen bir hata oluştu. Hata kodu:", e)
```
