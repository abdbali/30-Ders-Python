## 1. Hazır Gömülü (Built-in) Fonksiyonlar

Python açılır açılmaz emrine amade olan, `import` etmene gerek olmayan komutlardır. Zaten `print()`, `input()`, `len()`, `type()` gibi pek çoğunu kullandık. İşte matematikte çok işine yarayacak birkaç yenisi:

```python
sayilar_listesi = [15, 2, 85, 40, -10]

print("En büyük sayı:", max(sayilar_listesi))  # Maximum
print("En küçük sayı:", min(sayilar_listesi))  # Minimum
print("Tüm sayıların toplamı:", sum(sayilar_listesi))  # Sum (Toplam)

# abs() Mutlak değer demektir. Eksi bir sayıyı artıya çevirir (mesafe hesaplarında kullanılır)
sicaklik_farki = -20
print("Gerçek sıcaklık farkı:", abs(sicaklik_farki))
```

## 2. Modül (Kütüphane) Nedir ve import Mantığı

Her komut açık kalsaydı, bilgisayarın hafızası boş yere dolardı. Bu yüzden bazı özel alet takımları kapalı kutulardadır. Onları kullanmak için projemizin en başına **`import` (İçe Aktar)** yazarız.

```python
import math

# Artık math kütüphanesinin içindeki her şeye 'math.' yazarak ulaşabiliriz
print("Pi Sayısı:", math.pi)
print("16'nın Karekökü:", math.sqrt(16)) # sqrt = square root

ondalikli_sayi = 4.8
print("Aşağı Yuvarla:", math.floor(ondalikli_sayi)) # Zemin
print("Yukarı Yuvarla:", math.ceil(ondalikli_sayi)) # Tavan
```

## 3. Eğlenceli Kısım: random Kütüphanesi

Oyun yapımının kalbi rastgeleliktir. Düşmanın nereden çıkacağı, zarların kaç geleceği, şanslı çarkın nerede duracağı hep `random` (rastgele) kütüphanesiyle ayarlanır.

```python
import random

# 1 ile 10 arasında (10 dahil) rastgele tam sayı seçmek
sansli_sayi = random.randint(1, 10)
print("Şanslı Sayımız:", sansli_sayi)

# Listeden rastgele BİR eleman çekmek (Kura çekimi)
kisiler = ["Ali", "Ayşe", "Buse", "Kemal"]
kazanan = random.choice(kisiler)
print("Çekilişi Kazanan:", kazanan)

# Listeyi rastgele karıştırmak (Kart karmak gibi düşün)
print("Karıştırılmadan Önce:", kisiler)
random.shuffle(kisiler)
print("Karıştırıldıktan Sonra:", kisiler)
```
