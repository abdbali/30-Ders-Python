## 1. Metot (Method) Nedir?

Metot, aslında **Sınıfın (Class) içine yazılmış bir fonksiyondan (`def`) ibarettir.** 
Tek kuralı şudur: Sınıf içindeki her fonksiyon, parantezinin içine ilk parametre olarak `self` almak zorundadır. Çünkü eylemi yapacak olan bizzat nesnenin *kendisidir*.

```python
class Kedi:
    def __init__(self, isim):
        self.isim = isim
        
    # Kediye eylem (metot) ekliyoruz
    def miyavla(self):
        print(f"{self.isim} diyor ki: Miyavvv! Beni besle.")

# Kedimizi yaratalım
minnos = Kedi("Minnoş")

# Özelliği okumak (.isim)
print("Kedinin Adı:", minnos.isim)

# Eylemi (metodu) çağırmak (.miyavla())
# NOT: Metotlar her zaman () ile bitmelidir!
minnos.miyavla()
```

## 2. Metot İçinden Özelliği Güncellemek

Bir oyun karakteri iksir içtiğinde canının artması lazımdır. Biz de sınıfın içine `iksir_ic()` adında bir metot yazıp, bu metot çalıştığında o nesnenin KENDİ CANINI (`self.can`) 20 arttırmasını sağlayabiliriz.

```python
class Karakter:
    def __init__(self, ad):
        self.ad = ad
        self.can = 100 # Başlangıçta herkesin 100 canı var
        
    def iksir_ic(self):
        print(f"\n{self.ad} yeşil bir iksir içiyor...")
        self.can += 20
        print(f"Yeni Canı: {self.can}")
        
    def hasar_al(self, hasar_miktari):
        print(f"\nOlamaz! {self.ad} {hasar_miktari} hasar aldı.")
        self.can -= hasar_miktari
        if self.can <= 0:
            print(f"{self.ad} elendi...")
        else:
            print(f"Kalan Canı: {self.can}")


kahraman = Karakter("Arthur")
print("Oyun Başladı! Karakter:", kahraman.ad, "Can:", kahraman.can)

# Kahramana komutlar (metotlar) verelim
kahraman.hasar_al(40)
kahraman.iksir_ic()
kahraman.hasar_al(90)
```
