## 2. Dosyaya Yazmak ("w" Modu - Write)

Eğer olmayan bir dosyayı `w` (Write - Yaz) modunda açarsan, Python o dosyayı senin için **sıfırdan yaratır**. 
DİKKAT: Eğer dosya zaten varsa ve içinde yazılar varsa, `w` modu onu tamamen SİLER ve üzerine sıfırdan yazar!

```python
# 'benim_dosyam.txt' adında bir dosya oluşturuyoruz ve içine yazıyoruz
yeni_dosya = open("benim_dosyam.txt", "w", encoding="utf-8")

# encoding="utf-8" kısmı Türkçe karakterlerin (ş, ğ, ç, ö vb.) bozuk görünmesini engeller!

yeni_dosya.write("Merhaba Dünya!\n")
yeni_dosya.write("Bu benim ilk dosyam.\n")
yeni_dosya.close()

print("Dosya başarıyla oluşturuldu ve içine yazıldı! Klasörünü kontrol et.")
```

## 3. Dosyaya Ekleme Yapmak ("a" Modu - Append)

`w` modu eski yazıları siliyor demiştik. Bir oyuncunun her yeni skorunu eski skorların ALTINA eklemek istiyorsak, listelerdeki `append()` metodundan hatırlayacağın `a` (Append - Ekle) modunu kullanmalıyız.

```python
# Önceden oluşturduğumuz dosyayı "a" moduyla açıyoruz
ekleme_dosyasi = open("benim_dosyam.txt", "a", encoding="utf-8")

ekleme_dosyasi.write("Bu satır sonradan eklendi! Eski yazılar silinmedi.\n")
ekleme_dosyasi.close()

print("Ekleme işlemi başarılı.")
```

## 4. with open() Yapısı (Güvenli Yol)

Programcılar genellikle `.close()` yazmayı unuturlar ve program çöker. Bunu kökten çözmek için `with open(...) as dosya:` yapısı kullanılır. 
Bu yapının altındaki girintili (boşluklu) alan bittiğinde, Python dosyayı bizim yerimize OTOMATİK olarak kapatır.

```python
# Hem güvenli hem daha kısa okuma işlemi
with open("benim_dosyam.txt", "r", encoding="utf-8") as dosya:
    tum_yazilar = dosya.read()
    print("--- DOSYA OKUNUYOR ---")
    print(tum_yazilar)
    
# Bu satıra indiğimizde dosya çoktan otomatik kapatıldı bile!
```
