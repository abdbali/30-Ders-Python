## 1. Tuple Nedir ve Nasıl Oluşturulur?

Listeler köşeli parantez `[]` ile oluşturulurdu. Tuple'lar ise yuvarlak parantez `()` ile oluşturulur.
Eğer içindeki elemanları virgülle ayırıp parantez koymazsan bile Python onu otomatik olarak Tuple zanneder!

```python
# Yuvarlak parantez ile Tuple
haftanin_gunleri = ("Pazartesi", "Salı", "Çarşamba", "Perşembe", "Cuma", "Cumartesi", "Pazar")

# Koordinatlar için harika bir örnektir
hazine_koordinati = (41.015, 28.979)

print("Hazine Konumu:", hazine_koordinati)
print("Veri türü:", type(haftanin_gunleri))
```

## 2. Tuple İçinden Veri Okumak (İndeksleme)

Tuple'ın içindeki verilere ulaşmak listelerle TAMAMEN AYNIDIR. Köşeli parantez kullanarak 0. veya 1. indeksine ulaşabiliriz, çünkü sırası bellidir.

```python
renkler = ("Kırmızı", "Yeşil", "Mavi")

print("İlk renk:", renkler[0])
print("Son renk:", renkler[-1])
```

## 4. Tür Dönüşümleri (List <-> Tuple)

Diyelim ki sana bir Tuple verildi ama senin onu bir şekilde değiştirmen şart oldu. Ne yapacaksın? 
Önce onu listeye çevireceksin (`list()`), istediğin değişikliği yapacaksın, sonra tekrar güvenlik çemberine, yani Tuple'a alacaksın (`tuple()`).

```python
korumali_sifreler = ("sifre123", "qwerty")

# 1. Adım: Listeye Çevir
gecici_liste = list(korumali_sifreler)

# 2. Adım: Değişiklik Yap
gecici_liste.append("yeniSifre_99")

# 3. Adım: Tekrar Tuple yap (Güvenliğe al)
korumali_sifreler = tuple(gecici_liste)

print("Güncellenmiş Korumalı Şifreler:", korumali_sifreler)
```
