## 1. Liste Nasıl Oluşturulur?

Bir liste oluşturmak, değişken oluşturmaya çok benzer. Sadece koyacağımız birden fazla değeri **köşeli parantezler `[]`** içine alırız ve aralarına **virgül `,`** koyarız.

Bunu bir tren gibi düşün: Köşeli parantezler trenin başı ve sonudur, virgüller ise vagonları birbirine bağlayan kancalardır.

```python
# İçinde üç farklı isim barındıran bir liste
ogrenciler = ["Ali", "Ayşe", "Veli"]

# İçinde sayılar barındıran bir liste
sinav_notlari = [85, 90, 78, 100]

# Hiçbir şey içermeyen boş bir liste (İleride içini doldurmak üzere)
bos_envanter = []

print("Öğrenciler listesi:", ogrenciler)
print("Veri türü:", type(ogrenciler))
```

## 2. Listelerin İçindeki Verilere Ulaşmak (İndeksleme)

Gün 4'te metinlerin (String) içinden harf alırken bilgisayarların saymaya 0'dan başladığını öğrenmiştik. Aynı kural listeler için de geçerlidir! 
Bir listenin ilk elemanını istiyorsak, 0 numaralı vagonu çağırmalıyız.

```python
meyveler = ["Elma", "Muz", "Çilek", "Kivi"]
           #  0       1        2        3     (İndeks Numaraları)

print("Listenin ilk meyvesi:", meyveler[0])
print("Listenin üçüncü meyvesi:", meyveler[2])

# Tıpkı kelimelerde olduğu gibi eksi (-) kullanarak sondan da sayabiliriz!
print("Listenin en son meyvesi:", meyveler[-1])
```

## 3. Listelerin İçeriğini Değiştirmek

Listelerin en güzel yanı 'değiştirilebilir' (mutable) olmalarıdır. Bir metnin (String) ortasındaki bir harfi direkt değiştiremeyiz ama listelerdeki herhangi bir vagonun içindeki eşyayı çöpe atıp yerine yenisini koyabiliriz.

```python
oyun_envanteri = ["Tahta Kılıç", "Can İksiri", "Harita"]
print("Eski Envanter:", oyun_envanteri)

# Diyelim ki karakterimiz level atladı ve tahta kılıcı çöpe atıp yerine demir kılıç aldı.
# 0. indeksteki veriyi değiştiriyoruz:
oyun_envanteri[0] = "Demir Kılıç"

print("Yeni Envanter:", oyun_envanteri)
```

## 4. Farklı Veri Tiplerini Aynı Listede Tutmak

Python listeleri çok esnektir. Bir listenin içine aynı anda metin (String), tam sayı (Integer), ondalıklı sayı (Float) hatta `True/False` (Boolean) bile koyabilirsin.

```python
# Sırasıyla: İsim, Yaş, Boy, Öğrenci mi?
karisik_liste = ["Zeynep", 16, 1.65, True]

print(f"{karisik_liste[0]} isimli kişinin yaşı {karisik_liste[1]}'dır.")
```

## 5. Listeler ve for Döngüsü (Ayrılmaz İkili)

`for` döngüsünü hatırladın mı? Listeler ve `for` döngüleri programlamanın ayrılmaz ikilisidir. Bir listenin içindeki tüm vagonları tek tek gezmek için `for` kullanırız.

```python
renkler = ["Kırmızı", "Mavi", "Yeşil", "Sarı"]

print("Sahip olduğumuz renkler:")
for renk in renkler:
    print("-", renk)
```
