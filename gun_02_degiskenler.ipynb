## 1. Değişken Nedir ve Nasıl Oluşturulur?

Bir oyun yaptığımızı hayal edelim. Oyuncunun ismini saklamak için bir kutu alıyoruz, üzerine 'oyuncu_ismi' yazıyoruz ve içine 'Ahmet' değerini koyuyoruz. Python'da bunu yapmak çok basittir: Sadece etiketin adını yaz, eşittir (=) işareti koy ve kutunun içine ne koymak istediğini belirt.

Buradaki = işareti matematikteki 'eşittir' anlamına gelmez. Programlamada bu 'atama' işaretidir. Yani 'sağdaki değeri al, soldaki kutunun içine koy' demektir.

```python
# 'oyuncu_ismi' adında bir değişken oluşturduk ve içine 'Ahmet' metnini koyduk.
oyuncu_ismi = "Ahmet"

# 'oyuncu_puani' adında bir değişken oluşturduk ve içine 100 sayısını koyduk.
oyuncu_puani = 100

# Şimdi bu kutuların içindekileri ekrana yazdıralım:
print("Oyuncunun Adı:")
print(oyuncu_ismi)

print("Oyuncunun Başlangıç Puanı:")
print(oyuncu_puani)
```

4. Sadece harf, rakam ve alt çizgi (_) içerebilir.
5. Özel karakterler (!, @, #, $, %, vs.) kullanılamaz.

## 3. Temel Veri Tipleri (Data Types)

Kutularımızın içine farklı türden eşyalar koyabiliriz. Programlamada veri türünü bilmek çok önemlidir çünkü bir yazıyla bir sayıyı toplayamazsın.

Temel Veri Tipleri:
- String (str): Metin verileridir. Her zaman tırnak (" ") içinde yazılırlar. (Örn: "Merhaba", "Kılıç")
- Integer (int): Tam sayılardır. Küsuratı olmayan sayılardır. Tırnaksız yazılır. (Örn: 10, -5, 100)
- Float (float): Ondalıklı sayılardır. Türkçe'de biz virgül (3,14) kullansak da programlamada nokta (3.14) kullanılır. (Örn: 3.14, 2.5, -0.5)

```python
# String (Metin)
karakter_sinifi = "Büyücü"

# Integer (Tam Sayı)
seviye = 5

# Float (Ondalıklı Sayı)
hareket_hizi = 2.4

print("Karakter Sınıfı:", karakter_sinifi)
print("Seviyesi:", seviye)
print("Hareket Hızı:", hareket_hizi)
```

## 4. Değişkenin Türünü Öğrenmek: type() Fonksiyonu

Bazen bir kutunun içinde ne tür bir veri olduğunu bilemeyebiliriz. Bu durumda Python'a 'Bu değişkenin türü ne?' diye sorabiliriz. Bunun için 'type()' komutunu kullanırız.

```python
gizli_kutu = 42
diger_kutu = "42"
son_kutu = 42.0

print("gizli_kutu'nun türü:", type(gizli_kutu))
print("diger_kutu'nun türü:", type(diger_kutu))
print("son_kutu'nun türü:", type(son_kutu))
```

## 5. Değişkenlerin İçeriğini Değiştirmek ve Güncellemek

Değişkenlere neden değişken diyoruz? Çünkü içlerindeki değerler oyun veya program boyunca değişebilir! Bir kutunun içindeki eski eşyayı çöpe atıp yerine yenisini koyabiliriz. Veya içindeki sayıyı matematiksel işlemlerle artırabiliriz.

```python
puan = 0
print("Oyun başladı, puan:", puan)

# Oyuncu bir altın topladı!
puan = 10
print("Altın toplandı! Yeni puan:", puan)

# Oyuncu 5 puanlık bir düşman yendi. Mevcut puanın üzerine 5 ekleyelim.
puan = puan + 5
print("Düşman yenildi! Güncel puan:", puan)
```
