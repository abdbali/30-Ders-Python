# Gün 10: Döngülere Hükmetmek (break ve continue)

Hoş geldin! Bugün 10. günümüz ve Seviye 1'in son dersindeyiz! 
Önceki derslerde döngülerin (`while` ve `for`) nasıl çalıştığını öğrendik. Fakat bazen döngü 100 kere dönecek şekilde ayarlansa bile, aradığımız şeyi 5. turda bulabiliriz. Kalan 95 turu boşuna dönüp bilgisayarı yormanın alemi yoktur.

İşte döngüleri yarıda kesmek veya bazı adımları atlamak (pas geçmek) için iki özel komutumuz var: `break` ve `continue`.

**Günün Hedefleri:**
- `break` komutu ile döngüyü anında yok etmeyi (fren yapmayı) öğrenmek.
- `continue` komutu ile sadece o anki turu atlayıp döngüye devam etmeyi kavramak.
- `while True:` mantığıyla kontrollü sonsuz döngüler yaratıp onları `break` ile kırmayı görmek.

## 1. break Komutu (Freni Çekmek)

`break` (kırmak, parçalamak) komutu döngünün içindeyken okunduğu anda, döngü anında sonlanır. Döngünün daha kaç turu kaldığına bakılmaz, Python hemen döngünün dışına atlar.
Genellikle bir `if` şartının içine yazılır. "Eğer şu olursa, döngüyü durdur!" mantığıyla çalışır.

```python
print("Polis araması başladı...")

# 1'den 10'a kadar odaları arayacağız
for oda_no in range(1, 11):
    print(f"Oda {oda_no} aranıyor...")
    
    if oda_no == 4:
        print("Hırsız 4 numaralı odada bulundu! Arama iptal ediliyor.")
        break # break çalıştığı için 5, 6, 7.. odalara ASLA bakılmayacak.

print("Arama operasyonu bitti.")
```

## 2. continue Komutu (Pas Geçmek)

`continue` (devam et) komutu döngüyü tamamen bitirmez. SADECE o anki turu iptal eder ve döngüyü hemen bir sonraki turdan (başa dönerek) devam ettirir.

Bunu bir yolda yürürken önümüze çıkan su birikintisinden atlamak gibi düşünebilirsin. Yürümeyi bırakmıyorsun, sadece o adımı atlıyorsun.

```python
print("Haftanın günleri işleniyor...")

for gun in range(1, 8):
    # Hafta sonu günlerini (6 ve 7) pas geçmek istiyoruz
    if gun == 6 or gun == 7:
        print(f"{gun}. Gün hafta sonu! Çatışma yok, dinleniyoruz.")
        continue # Continue altındaki kodları okumaz, hemen döngünün başına fırlar!
        
    print(f"{gun}. Gün işe gidildi ve çalışıldı.")
```

## 3. while True: ve break Birlikteliği

Gün 8'de sonsuz döngülerin zararlı olduğunu söylemiştik. Ancak bazen bilerek ve isteyerek sonsuz döngü yaratırız: `while True:` yazarak!

Peki program çökmesin diye ne yapacağız? İçine bir `if` ve `break` koyacağız. Kullanıcı doğru bir şey yapana kadar döngü dönecek, doğru yapınca `break` ile kıracağız. En yaygın menü ve oyun motoru taktiği budur.

```python
print("--- Gizli Kasaya Hoş Geldiniz ---")

while True: # Burası her zaman True olduğu için sonsuza kadar döner!
    cevap = input("Kasayı açmak için parolayı girin (Çıkmak için 'q'): ")
    
    if cevap == "q":
        print("Sistem kapatılıyor...")
        break # Döngü kırıldı!
        
    elif cevap == "Python2024":
        print("Kasa açıldı! Zengin oldun!")
        break # Döngü kırıldı!
        
    else:
        print("Yanlış parola! Sistem tekrar soracak...\n")
        # Burada break yok, döngü başa dönüp tekrar soracak.
```
