## 1. Set (Küme) Nedir ve Nasıl Oluşturulur?

Kümeler **süslü parantez `{}`** ile oluşturulur. İki büyük kuralı vardır:
1. **Benzersizdir:** Aynı elemandan asla iki tane bulunamaz.
2. **Sırasızdır:** Elemanların belli bir sırası (0., 1. indeks) YOKTUR. Bu yüzden `kume[0]` diye bir kodu çalıştıramazsın.

```python
# Süslü parantez ile Set oluşturalım
farkli_renkler = {"Mavi", "Kırmızı", "Yeşil"}

# Kopya elemanlar vererek bir set oluşturalım. Bakalım ne olacak?
kopya_denemesi = {"Elma", "Muz", "Elma", "Çilek", "Muz"}

print("Kopya Denemesi Sonucu:", kopya_denemesi)
# Gördüğün gibi Python aynı olanları anında sildi, geriye sadece 1'er tane bıraktı!
```

## 2. Listeden Kopyaları Temizlemenin En Hızlı Yolu

Diyelim ki elinde binlerce kelimelik bir liste var ve içinde bir sürü kopya (aynı) kelime var. Bunları tek tek `for` döngüsüyle bulup silmek çok yavaştır. 
Bunun yerine o listeyi Set'e çevirip, sonra tekrar Listeye çevirmek sadece 1 saniye sürer!

```python
kirli_liste = ["Kitap", "Kalem", "Kitap", "Silgi", "Kalem", "Kalem"]
print("Eski Liste:", kirli_liste)

# 1. Set'e çevir (kopyalar otomatik silinir)
temiz_set = set(kirli_liste)

# 2. Tekrar listeye çevir
temiz_liste = list(temiz_set)

print("Temizlenmiş Liste:", temiz_liste)
```

## 3. Set'e Eleman Eklemek ve Çıkarrmak

Listelerde `.append()` kullanıyorduk, Set'lerde ise `.add()` kullanılır. 
Silmek için ise yine `.remove()` kullanılır.

```python
sayilar_kumesi = {10, 20, 30}

# Yeni eleman ekleyelim
sayilar_kumesi.add(40)

# Zaten var olan bir şeyi eklemeye çalışalım (Hiçbir şey olmaz, hata vermez ama eklemez de)
sayilar_kumesi.add(20)

# 10 sayısını silelim
sayilar_kumesi.remove(10)

print("Güncel Küme:", sayilar_kumesi)
```

## 4. Kümelerde Matematik (Kesişim, Birleşim, Fark)

Venn şemalarını hatırla. İki farklı kümenin ortak noktalarını veya farklı noktalarını bulmak Python'da çok kolaydır.
- Kesişim (Ortak): `.intersection()` veya `&` işareti.
- Birleşim (Tümü): `.union()` veya `|` işareti.
- Fark (Sadece birinde olan): `.difference()` veya `-` işareti.

```python
satranc_kulubu = {"Ali", "Ayşe", "Veli", "Zeynep"}
robotik_kulubu = {"Ayşe", "Mehmet", "Zeynep", "Can"}

# Her iki kulübe de giden zeki öğrenciler kimler?
ortaklar = satranc_kulubu.intersection(robotik_kulubu)
print("Her İki Kulüpte Olanlar:", ortaklar)

# Okuldaki tüm aktif öğrencilerin toplamı (Kopya isimler teke düşer)
tum_ogrenciler = satranc_kulubu.union(robotik_kulubu)
print("Tüm Aktif Öğrenciler:", tum_ogrenciler)

# Sadece satranç kulübünde olup robotikte olmayanlar
sadece_satranc = satranc_kulubu.difference(robotik_kulubu)
print("Sadece Satranca Gidenler:", sadece_satranc)
```
