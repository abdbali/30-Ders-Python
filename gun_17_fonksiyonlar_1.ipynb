## 1. Fonksiyon Nasıl Tanımlanır ve Çağrılır?

Fonksiyon tanımlamak (üretmek) için İngilizce 'define' (tanımla) kelimesinin kısaltması olan **`def`** anahtar kelimesini kullanırız.

Kural: `def fonksiyon_adi():` yazarız, altına (girintili olarak) fonksiyonun ne yapacağını yazarız. 
ÖNEMLİ: Fonksiyonu tanımlamak, onu çalıştırmak demek DEĞİLDİR. Sadece bir makine üretmiş oluruz. Çalıştırmak için adını çağırmamız gerekir.

```python
# 1. Adım: Fonksiyonu Tanımla (Makineyi üret)
def selamla():
    print("Merhaba! Sisteme hoş geldiniz.")
    print("Bugün nasılsınız?")
    
# Yukarıdaki kod çalışınca ekrana HİÇBİR ŞEY yazmaz, çünkü sadece makineyi ürettik.

# 2. Adım: Fonksiyonu Çağır (Makineyi çalıştır)
print("--- Fonksiyon Çağrılıyor ---")
selamla()
selamla() # İstediğimiz kadar kullanabiliriz!
```

## 2. Parametre Alan Fonksiyonlar (Dışarıdan Malzeme Almak)

Yukarıdaki `selamla()` çok kısıtlıdır, herkese aynı mesajı verir. Oysa biz `print("Ali")` dediğimizde Ali, `print("Ayşe")` dediğimizde Ayşe yazar. Biz de kendi fonksiyonumuzun parantezleri arasına **parametreler (değişkenler)** koyabiliriz.

Bu tıpkı bir meyve suyu makinesine (fonksiyon) elma atarsan (parametre) elma suyu, portakal atarsan portakal suyu çıkması gibidir.

```python
# isim adında bir parametre (malzeme) bekliyor
def ozel_selamla(isim):
    print(f"Merhaba {isim}! Sisteme hoş geldin.")

# Çağırırken parantez içine malzemeyi veriyoruz
ozel_selamla("Kaan")
ozel_selamla("Zeynep")

# Birden fazla parametre de alabiliriz
def toplama_yap(sayi1, sayi2):
    sonuc = sayi1 + sayi2
    print(f"{sayi1} + {sayi2} = {sonuc}")

toplama_yap(15, 20)
toplama_yap(100, 50)
```

## 3. Neden Fonksiyon Yazarız? (DRY Prensibi)

Yazılım dünyasının en ünlü kurallarından biri **DRY (Don't Repeat Yourself - Kendini Tekrar Etme)** prensibidir. Eğer aynı veya çok benzer kodları 3-4 farklı yere kopyalayıp yapıştırıyorsan, orada bir hata yapıyorsun demektir. Hemen o kodu bir `def` içine alıp fonksiyon yapmalısın.

## 4. Varsayılan (Default) Parametreler

Bazen fonksiyonu çağıran kişi parantez içini boş bırakabilir. Hata vermemesi için, parametreye `isim="Misafir"` şeklinde yedek bir varsayılan değer atayabiliriz.

```python
# Eğer kisi parametresi gönderilmezse, otomatik olarak "Gizemli Yolcu" olur
def ucaga_bindir(kisi="Gizemli Yolcu"):
    print(f"{kisi} uçağa biniş yaptı.")

ucaga_bindir("Mustafa") # Parametre verdik, Mustafa yazar
ucaga_bindir()          # Boş bıraktık, yedek parametre olan Gizemli Yolcu yazar!
```
