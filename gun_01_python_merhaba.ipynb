## 1. print() Fonksiyonu ile İlk İletişim

Günlük hayatta büyük bir kalabalığa seslenmek istesen ne kullanırsın? Bir megafon, değil mi? 
Python'da da bilgisayar ekranına (konsola) bir şeyler söylemek, yani yazdırmak istediğimizde 'print()' komutunu kullanırız. 'print' İngilizcede 'yazdır' demektir.

Parantezlerin () içini megafonun içine koyacağımız mesaj, tırnak işaretlerini "" ise söyleyeceğimiz cümlenin başlangıcı ve bitişi olarak düşünebilirsin. Bilgisayar sadece tırnak içindeki kısmı okuyacak ve ekrana yansıtacaktır.

```python
# Bilgisayara ilk komutumuzu veriyoruz!
print("Merhaba Dünya, ben harika projeler yapmaya geldim!")
```

## 2. Metinler (Stringler) ve Sayılar (Integer)

Programlama dilleri veri tiplerine çok önem verir. Python için yazılan bir şeyin yazı mı yoksa sayı mı olduğu çok önemlidir. 

Eğer bir metin yazmak istiyorsan, bunu MUTLAKA tırnak içinde yazmalısın. Python tırnak içindeki her şeyi 'Bu bir yazıdır, içinde matematiksel bir işlem bile olsa bunu olduğu gibi ekrana basacağım' şeklinde algılar.

Fakat eğer gerçekten matematiksel bir işlem yaptırmak istiyorsan, tırnak işareti KULLANMAMALISIN. Aşağıdaki örneği dikkatlice incele:

```python
# Bu sadece bir metindir, ekrana 5 + 3 yazar.
print("5 + 3")

# Bu ise gerçek bir matematik işlemidir, Python bunu hesaplar ve ekrana 8 yazar.
print(5 + 3)

# Başka işlemler de yapabiliriz
print(10 - 4)
print(10 * 2)  # Yıldız işareti çarpma işlemi içindir
print(20 / 5)  # Eğik çizgi bölme işlemi içindir
```

## 3. print() İçinde Birden Fazla Değer Yazdırmak

Bazen hem bir metin hem de bir sayıyı aynı satırda yazdırmak isteyebiliriz. print komutunun içine birden fazla şey koymak için, aralarına virgül (,) koymamız yeterlidir. Python bu parçaları otomatik olarak aralarında bir boşluk bırakarak birleştirir.

```python
print("Benim yaşadığım yıl:", 2024)
print("Oyun yükleniyor... Yüzde:", 99)
print("Karakterin Adı:", "Savaşçı", "- Karakterin Seviyesi:", 10)
```

## 7. Günün Projesi: ASCII Sanatı

Bilgisayarlar ilk icat edildiğinde gelişmiş grafikler, renkli pikseller yoktu. İnsanlar sadece klavyedeki karakterleri kullanarak ekranda çizimler yapardı. Buna ASCII sanatı denir. Biz de bugün öğrendiğimiz print komutunu alt alta kullanarak kendi sanat eserimizi oluşturacağız.

```python
# Örnek: Basit bir kedi yüzü çizimi
print(" /\_/\ ")
print("( o.o )")
print(" > ^ < ")
print("")

# Örnek: Bir araba çizimi
print("  ______")
print(" /|_||_\`.__")
print("(   _    _ _\")
print("=`-(_)--(_)-' ")
print("")

# ŞİMDİ SIRA SENDE: Kendi istediğin bir şeyi (Ev, kılıç, robot, harf) alt alta print komutları kullanarak çiz!
print(" ")
print(" ")
print(" ")
print(" ")
```
