# Gün 24: Sınırsız Parametreler (*args ve **kwargs)

Hoş geldin! Fonksiyonlar konusunda (`def`) dışarıdan değişkenler (parametre) almayı öğrenmiştik: `def topla(a, b):`. 
Ancak bu fonksiyona sadece 2 sayı gönderebiliriz. Kullanıcı 5 sayıyı aynı anda toplamak isterse ne olacak? Fonksiyon çökecektir.

İşte Python'da **"Sınırları kaldırmak ve istediğin kadar çok veri almak"** için iki efsanevi özellik vardır: `*args` ve `**kwargs`.

**Günün Hedefleri:**
- `*args` kelimesinin başındaki yıldızın (`*`) sınırsız değişken (Tuple olarak) alma gücünü anlamak.
- `**kwargs` ile sınırsız Anahtar-Değer (Sözlük olarak) verisi almayı kavramak.
- Esnek, çökmeyen ve her duruma uyum sağlayan fonksiyonlar yazmak.

## 1. *args Nedir? (Sınırsız Argümanlar)

Parametrenin başına tek bir yıldız `*` koyarsak (İsmi *args olmak zorunda değildir, *sayilar veya *isimler de olabilir), Python fonksiyona gönderilen tüm ekstra verileri bir **Tuple (Demet)** içinde toplayıp o değişkene atar.

```python
# Ne kadar sayı geleceğini bilmiyoruz, başına * koyduk
def limitsiz_topla(*sayilar):
    print("Gelen Sayılar (Tuple formatında):", sayilar)
    
    toplam = 0
    # Bir Tuple olduğu için for döngüsü ile içindekileri tek tek toplayabiliriz
    for sayi in sayilar:
        toplam += sayi
        
    return toplam

# İstediğin kadar parametre gönder, hata vermez!
sonuc1 = limitsiz_topla(10, 20, 30)
sonuc2 = limitsiz_topla(5, 1, 2, 8, 9, 10, 11)

print("İlk Toplam:", sonuc1)
print("İkinci Toplam:", sonuc2)
```

## 2. **kwargs Nedir? (Sınırsız İsimli Argümanlar)

İngilizce 'Keyword Arguments' kelimesinden gelir. İki yıldız `**` kullanırsak, Python bize gönderilen verileri bir **Sözlük (Dictionary)** olarak paketler.
Bunu kullanmak için fonksiyona veri yollarken `anahtar="değer"` formatında yollamamız gerekir.

```python
# Kullanıcının profiline sınırsız sayıda özellik eklemesine izin verelim
def profil_olustur(**ozellikler):
    print("Gelen Sözlük (Dict):", ozellikler)
    
    print("--- Oyuncu Profili ---")
    for anahtar, deger in ozellikler.items():
        print(f"{anahtar}: {deger}")

# Sınırsız sayıda anahtar=değer şeklinde veri gönderebiliriz
profil_olustur(isim="Mete", sinif="Okçu", seviye=45, klan="Ejderler")
```

## 3. Sabit Parametreler ile Birlikte Kullanmak

Fonksiyona her zaman sabit olarak alınması GEREKEN parametreler koyabilir, 'geri kalan fazlalıkları' `*args` içine toplayabiliriz.
Kural: Önce her zaman normal parametreler, sonra `*args`, en son `**kwargs` yazılır.

```python
# isim ZORUNLUDUR. Yanına gelen ek özellikler *yetenekler kutusuna düşer.
def karakter_tanit(isim, *yetenekler):
    print(f"Karakterin Adı: {isim}")
    print("Yetenekleri:")
    for yetenek in yetenekler:
        print("-", yetenek)

karakter_tanit("Büyücü Galdor", "Ateş Topu", "Görünmezlik", "Işınlanma")
```
