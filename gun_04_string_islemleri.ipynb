## 1. String Birleştirme (Concatenation)

Artı (+) işaretini sayılarda kullandığımızda toplama işlemi yaptığını biliyoruz. Peki ya iki metni (string) toplarsak ne olur? 
Cevap: Python o iki metni yapıştırarak tek bir metin haline getirir. Ancak dikkat et, boşlukları otomatik eklemez!

```python
isim = "Ahmet"
soyad = "Yılmaz"

# Doğrudan toplarsak yapışık olurlar
tam_isim_yapisik = isim + soyad
print("Yapışık İsim:", tam_isim_yapisik)

# Araya boşluk karakteri ekleyerek toplamalıyız
tam_isim_duzgun = isim + " " + soyad
print("Düzgün İsim:", tam_isim_duzgun)

# Metinlerle sayıları DOĞRUDAN toplayamayız! (Hata verir)
# print(isim + 5) # Bu kodu çalıştırırsan TypeError hatası alırsın.
```

## 2. String Çoğaltma

Çarpı (*) işaretini bir metin ve bir sayıyla kullanırsak, Python o metni sayı kadar yan yana yazar.

```python
gulucuk = "Ha"
print("Komik bir şey oldu:", gulucuk * 5)

ayirac = "-"
print(ayirac * 30)
print("Oyun Bitti")
print(ayirac * 30)
```

## 3. String Formatlama (f-strings) - EN ÖNEMLİ KONU!

Metinlerin arasına değişken koymak için artı işareti kullanmak uzun ve yorucu bir işlemdir. Üstelik metinle sayıyı direkt toplayamadığımız için işler karışır.

Python'da bunun çok kolay ve modern bir yolu vardır: **f-strings** (format string).
Tırnak işaretinin hemen başına 'f' harfi koyarsak, tırnağın içindeyken süslü parantez {} açarak değişkenlerimizi metnin içine sihirli bir şekilde yerleştirebiliriz.

```python
kahraman_ismi = "Leon"
seviye = 45
silah = "Buz Kılıcı"

# Eski ve zor yöntem (Artı işareti ile):
# print("Kahramanımız " + kahraman_ismi + ", şu an " + str(seviye) + " seviyesinde.")

# Yeni ve güçlü f-string yöntemi:
mesaj = f"Kahramanımız {kahraman_ismi}, şu an {seviye} seviyesinde ve elinde {silah} var!"
print(mesaj)
```

## 4. Metin Uzunluğunu Bulmak: len() Fonksiyonu

Bir kelimede kaç harf olduğunu, ya da bir şifrenin kaç karakter uzunluğunda olduğunu bulmak için 'len()' (İngilizce length yani uzunluk kelimesinin kısaltması) komutunu kullanırız. Boşlukların da bir karakter olduğunu unutma!

```python
parola = "GizliSifrem123"
cumle = "Merhaba Dünya"

print("Parola uzunluğu:", len(parola))
print("Cümle uzunluğu (Boşluk dahil):", len(cumle))
```

## 5. String Metotları (Yazıları Düzenlemek)

String veri tipinin kendine ait gizli güçleri (metotları) vardır. Bir string'in yanına nokta (.) koyarak bu yetenekleri çağırabiliriz.

- upper(): Tüm harfleri büyük yapar.
- lower(): Tüm harfleri küçük yapar.
- capitalize(): Sadece ilk harfi büyük yapar.
- title(): Her kelimenin ilk harfini büyük yapar.

```python
isim_soyisim = "ali veli"
bagirma = "HEY BURAYA BAKIN!"

print("Orijinal:", isim_soyisim)
print("Büyük Harf:", isim_soyisim.upper())
print("Baş Harfi Büyük:", isim_soyisim.capitalize())
print("Her Kelime Büyük:", isim_soyisim.title())
print("Küçük Harf (Fısıldama):", bagirma.lower())
```

## 6. İndeksleme: Bilgisayarlar Saymaya 0'dan Başlar!

Bir kelimenin içinden sadece tek bir harfi almak isteyebiliriz. Python'da kelimedeki her harfin bir sıra numarası (indeks) vardır.
ANCAK ÇOK ÖNEMLİ BİR KURAL: Programlamada saymaya 1'den değil, **0'dan** başlanır!

'P Y T H O N'
 0 1 2 3 4 5  (Sıra Numaraları)

```python
kelime = "PYTHON"

print("İlk harf (İndeks 0):", kelime[0])
print("İkinci harf (İndeks 1):", kelime[1])
print("Son harf (İndeks 5):", kelime[5])

# Negatif indeksleme kullanarak sondan da sayabiliriz. Sondaki harf her zaman -1'dir.
print("Sondan ilk harf (-1):", kelime[-1])
print("Sondan ikinci harf (-2):", kelime[-2])
```

## 7. Parçalama (Slicing)

Kelimenin içinden birden fazla harf kesip almak için araya iki nokta üst üste (:) koyarız.
Yazım kuralı: `kelime[başlangıç_indeksi : bitiş_indeksi]`
Not: Başlangıç indeksi dahildir, fakat bitiş indeksi dahil DEĞİLDİR.

```python
uzun_kelime = "PROGRAMLAMA"

print("Baştan 4. harfe kadar (0'dan 3'e dahil):", uzun_kelime[0:4])

# Başlangıcı boş bırakırsak en baştan alır, bitişi boş bırakırsak en sona kadar alır.
print("İlk 3 harf:", uzun_kelime[:3])
print("4. harften sona kadar:", uzun_kelime[4:])
```
