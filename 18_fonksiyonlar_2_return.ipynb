Örneğin, `len("Python")` komutu ekrana "6" yazdırıp kaybolmaz. Bize `6` sayısını bir hediye (ürün) olarak **geri verir (döndürür)**. Biz de bu `6` sayısını alır, başka bir matematik işleminde kullanırız.

İşte bugün, fonksiyonlarımızın sadece konuşmasını (`print`) değil, bize gerçek veriler, hesaplamalar teslim etmesini sağlayacak **`return` (Döndür)** anahtar kelimesini öğreneceğiz.

**Günün Hedefleri:**
- `print()` ve `return` arasındaki o devasa farkı anlamak.
- `return` kullanarak fonksiyonlardan dışarıya veri çıkartmak (değişkene atamak).
- Bir fonksiyondan aynı anda birden fazla veri (Tuple) döndürmek.

## 1. print() ve return Arasındaki Hayati Fark

Bunu bir pizzacı (fonksiyon) gibi düşün:
- `print()`: Pizzacı sana camın arkasından pizzayı gösterir, "Al, çok güzel oldu!" der ama pizzayı sana **VERMEZ**. Sadece bakarsın.
- `return`: Pizzacı pizzayı bir kutuya koyar ve dışarıya, **SENİN ELİNE** verir. Sen de o pizzayı alıp evine (bir değişkene) götürebilirsin.

```python
# print kullanan, sana bir şey teslim etmeyen fonksiyon (Sadece gösteriş)
def karesini_al_print(sayi):
    sonuc = sayi * sayi
    print("Karesi:", sonuc)

# return kullanan, sonucu sana teslim eden fonksiyon (Gerçek iş)
def karesini_al_return(sayi):
    sonuc = sayi * sayi
    return sonuc # Sonucu dışarı fırlat! Ekrana YAZDIRMAZ, teslim eder.

# TEST EDELİM:
x = karesini_al_print(5) 
# x değişkeni şu an BOŞ! Çünkü fonksiyon print yaptı, dışarı hiçbir şey fırlatmadı.

y = karesini_al_return(5)
# y değişkeni şu an 25 oldu! Fonksiyon bize pizzayı teslim etti.

print("Y'nin içi:", y)
```

## 2. return Nasıl Çalışır ve Neden Önemlidir?

Oyun programlıyorsan, oyuncunun yeni canını hesaplayan bir fonksiyon sadece ekrana "Yeni canınız 80" yazdırırsa oyun çöker! Oyun motoru o 80 sayısına ihtiyaç duyar. Bunu ancak `return` ile sağlayabiliriz.

**ÖNEMLİ KURAL:** Bir fonksiyon `return` satırını okuduğu an, **çalışmasını anında durdurur** ve dışarı çıkar. `return`'ün altına yazdığın hiçbir kod ASLA çalışmaz. Tıpkı döngülerdeki `break` gibidir!

```python
def yas_hesapla(dogum_yili, su_anki_yil=2024):
    yas = su_anki_yil - dogum_yili
    return yas
    print("Bu satır ASLA ama ASLA ekranda görünmeyecek çünkü return'ün altında!")

# Fonksiyondan gelen değeri alıp 'benim_yasim' kutusuna koyduk
benim_yasim = yas_hesapla(2005)
print(f"Sen tam {benim_yasim} yaşındasın.")
```

## 3. Birden Fazla Değer Döndürmek (Tuple Mucizesi)

Başka programlama dillerinde bir fonksiyondan sadece tek bir değer çıkarabilirsin. Ama Python çok esnektir! Bir fonksiyondan araya virgül koyarak 2, 3 hatta 10 farklı değeri aynı anda dışarı çıkarabilirsin (Python onları gizlice Tuple yapar).

```python
def islem_merkezi(a, b):
    toplam = a + b
    fark = a - b
    
    # İki sonucu birden dışarı yolluyoruz
    return toplam, fark

# Gelen iki sonucu, iki farklı değişkene tek satırda atıyoruz!
gelen_toplam, gelen_fark = islem_merkezi(20, 5)

print("Toplam:", gelen_toplam)
print("Fark:", gelen_fark)
```

## 6. Özet ve Bir Sonraki Güne Hazırlık

**Bugün Neler Öğrendik?**
- `print()`'in sadece ekrana yazı yazdığını ama kodun geri kalanına veri vermediğini.
- `return`'ün ise üretilen veriyi dışarı çıkartıp bizim elimize (bir değişkene) teslim ettiğini.
- `return` satırından sonra yazılan kodların asla çalışmayacağını (fonksiyonun bittiğini).
- Aralarına virgül koyarak birden fazla değeri Tuple şeklinde döndürebileceğimizi.

**Yarın Ne Yapacağız?**
Bazen bir değişken tanımlarız ama o değişkeni fonksiyonun içinden okuyamayız, ya da fonksiyonda oluşturduğumuz bir değişkene dışarıdan ulaşamayız. Sanki aralarında görünmez bir duvar vardır!
Yarın, bu duvarların ne olduğunu, **Kapsam (Local ve Global Scope)** kurallarını öğreneceğiz!
