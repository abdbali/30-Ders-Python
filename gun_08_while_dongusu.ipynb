## 1. while (Olduğu Sürece) Döngüsü Nasıl Çalışır?

`while` kelimesi İngilizcede 'iken', 'olduğu sürece' demektir. Mantığı `if` bloğuna çok benzer. Yanına bir şart yazarız.
Tek farkı: `if` bloğu şart doğruysa kodunu SADECE BİR KERE çalıştırır ve yoluna devam eder. `while` ise şart doğruysa kodunu çalıştırır, sonra **başa döner**, şartı tekrar kontrol eder. Şart False (Yanlış) olana kadar içindeki kodları döndürür durur.

```python
adim = 1

# Şartımız: adim değişkeni 5'ten küçük veya 5'e eşit OLDUĞU SÜRECE dön!
while adim <= 5:
    print(f"{adim}. adım atıldı.")
    # Çok önemli: Döngünün bir noktada bitmesi için adimi her turda 1 artırmalıyız.
    adim += 1  

print("Döngü bitti! Karakter hedefe ulaştı.")
```

## 3. Döngü İçinde Sayaç (Counter) Kullanımı

Döngüleri kontrol etmenin en güvenli yolu bir sayaç değişkeni kullanmaktır. Döngüden önce sayacı sıfırlar, her döngü turunda sayacı artırır veya azaltırız.

```python
# Geri sayım sayacı yapalım
roket_sayaci = 10

print("Roket fırlatma işlemi başlatılıyor...")

while roket_sayaci > 0:
    print(f"Fırlatmaya {roket_sayaci} saniye kaldı...")
    roket_sayaci -= 1  # Sayacı her turda 1 azaltıyoruz (Aksi halde sonsuz döngü olur!)

print("ATEŞ! 🚀")
```

## 4. Kullanıcıdan Sürekli Veri İstemek (Oyun Döngüsü Mantığı)

`while` döngüsünün en büyük gücü, kullanıcının doğru şeyi yapmasını beklemektir. Örneğin bir ATM'ye şifrenizi yanlış girdiğinizde kartı hemen yutmaz, "Tekrar deneyin" diyerek size bir hak daha sunar. İşte bu arka planda bir `while` döngüsüdür.

```python
gizli_sifre = "Python2024"
girilen_sifre = ""

# Girilen şifre doğru şifreye EŞİT OLMADIĞI SÜRECE döngü devam etsin
while girilen_sifre != gizli_sifre:
    girilen_sifre = input("Lütfen şifreyi giriniz (İpucu: Python2024): ")
    
    if girilen_sifre != gizli_sifre:
        print("Hatalı şifre, lütfen tekrar deneyin.\n")

# Döngüden sadece şifre doğruysa çıkılabilir!
print("Şifre doğru! Sisteme hoş geldiniz.")
```
