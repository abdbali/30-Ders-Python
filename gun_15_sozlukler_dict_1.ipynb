## 1. Sözlük (Dictionary) Nedir ve Nasıl Kurulur?

Sözlükler tıpkı Set'ler gibi **süslü parantez `{}`** ile oluşturulur. Ancak içindeki veriler tek başına değil, daima bir partnerle (ikili olarak) yazılır:
`"Anahtar" : "Değer"` (Key : Value)

Örneğin bir öğrenci sözlüğü oluşturalım.

```python
# Anahtar: İsim, Değer: Zeynep
# Anahtar: Yas, Değer: 15

ogrenci_karti = {
    "isim": "Zeynep",
    "yas": 15,
    "okul": "Anadolu Lisesi",
    "mezun_mu": False
}

print("Öğrenci Sözlüğü:", ogrenci_karti)
print("Veri türü:", type(ogrenci_karti))
```

## 2. Sözlükten Veri Okumak (Anahtarı Kullanmak)

Listelerde içindeki veriye ulaşmak için indeks numarasını `liste[0]` kullanıyorduk. 
Sözlüklerde sıra (indeks) YOKTUR. Bunun yerine etiketin kendisini, yani **Anahtarı (Key)** kullanırız. Tıpkı kasayı açmak için doğru anahtarı deliğe sokmak gibi!

```python
oyuncu_profili = {
    "kullanici_adi": "EjderAvcisi99",
    "seviye": 42,
    "sinif": "Savaşçı"
}

# Sadece seviyesini öğrenmek istiyoruz
print("Oyuncunun Seviyesi:", oyuncu_profili["seviye"])

# Sadece sınıfını öğrenmek istiyoruz
print("Oyuncunun Sınıfı:", oyuncu_profili["sinif"])
```

## 3. KeyError (Anahtar Hatası)

Eğer sözlükte olmayan (tanımlanmamış) bir anahtarı istersen Python kızar ve sana `KeyError` verir.
Bunu önlemek için güvence yöntemi olan `.get()` metodunu da kullanabiliriz.

```python
telefon_rehberi = {"Ali": "555-1234", "Ayşe": "555-9876"}

# HATA VEREN KULLANIM:
# print(telefon_rehberi["Veli"]) # KeyError verir çünkü Veli diye bir anahtar yok!

# GÜVENLİ KULLANIM:
veli_no = telefon_rehberi.get("Veli", "Kişi bulunamadı!")
print("Veli'nin Numarası:", veli_no)
```

## 4. Yeni Veri Eklemek veya Var Olanı Güncellemek

Sözlüğe sonradan yeni bir Anahtar-Değer ikilisi eklemek çok kolaydır. Sadece yeni bir anahtar yazıp ona değer atarız.
Eğer yazdığımız anahtar zaten varsa, içindeki değeri eskisini silip YENİLER.

```python
araba_ozellikleri = {
    "marka": "Ford",
    "renk": "Kırmızı"
}

print("Eski Hal:", araba_ozellikleri)

# 1. YENİ BİLGİ EKLEME (Yıl)
araba_ozellikleri["yil"] = 2020

# 2. VAR OLAN BİLGİYİ GÜNCELLEME (Rengi Mavi yapalım)
araba_ozellikleri["renk"] = "Mavi"

print("Yeni Hal:", araba_ozellikleri)
```
