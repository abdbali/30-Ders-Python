**Günün Hedefleri:**

## 1. Tüm Anahtarları veya Değerleri Görmek

Bir sözlükte ne tür bilgilerin saklandığını görmek için `.keys()` (anahtarlar) metodunu, içindeki cevapları liste halinde görmek için `.values()` (değerler) metodunu kullanırız.

```python
araba = {
    "marka": "Toyota",
    "model": "Corolla",
    "yil": 2022
}

print("Sözlükteki Etiketler (Anahtarlar):", araba.keys())
print("Sözlükteki Veriler (Değerler):", araba.values())
```

## 2. Sözlükleri for Döngüsüne Sokmak (.items())

Eğer bir sözlüğü direkt `for i in araba:` şeklinde döngüye sokarsan, Python sana sadece "Anahtarları" (marka, model, yil) verir, değerleri vermez.

Hem anahtarı hem de değeri aynı anda almak için çok özel bir metot kullanırız: `.items()`
Bu sayede `for` döngüsüne 2 değişken birden (k, v) verebiliriz.

```python
oyuncu_skorlari = {
    "Ali": 1500,
    "Ayşe": 2400,
    "Can": 900
}

# isim (key) ve skor (value) olarak iki değişken atıyoruz
print("--- Liderlik Tablosu ---")
for isim, skor in oyuncu_skorlari.items():
    print(f"Oyuncu: {isim} | Puanı: {skor}")
```

## 3. İç İçe Sözlükler (Nested Dictionaries)

Bazen bir değer, kendi içinde başka bir sözlük olabilir! 
Örneğin bir okulda birden fazla öğrenci var. Her öğrencinin adı bir 'Anahtar', o öğrencinin bilgileri ise bir 'Değer' (Sözlük) olabilir. İnternetten çektiğimiz (API) tüm veriler tam olarak bu formattadır.

```python
sinif_listesi = {
    "ogr_1": {"isim": "Ahmet", "yas": 14, "notu": 85},
    "ogr_2": {"isim": "Zeynep", "yas": 15, "notu": 95}
}

# 2 Numaralı öğrencinin sözlüğüne ulaşalım
print("Öğrenci 2 Bilgileri:", sinif_listesi["ogr_2"])

# 2 Numaralı öğrencinin sadece YAŞINA ulaşalım (İki defa köşeli parantez!)
print("Öğrenci 2'nin Yaşı:", sinif_listesi["ogr_2"]["yas"])
```

## 4. Sözlükten Eleman Silmek

Listelerde `.pop()` metodu indeks numarasıyla siliyordu. Sözlüklerde ise `.pop("Anahtar_Adi")` diyerek, o anahtarı ve ona bağlı değeri tamamen yok edebiliriz.

```python
telefon = {
    "marka": "Apple",
    "model": "iPhone 15",
    "renk": "Siyah",
    "fiyat": 50000
}

# Fiyat bilgisini sözlükten tamamen silelim
telefon.pop("fiyat")

print("Güncel Sözlük:", telefon)
```
