## 1. for Döngüsü ve range() Fonksiyonu

`for` kelimesi İngilizcede 'için' anlamına gelse de, programlamada 'her bir eleman için şunu yap' anlamı taşır.

Peki döneceğimiz elemanları nereden bulacağız? İşte burada `range()` (Aralık) fonksiyonu devreye girer. `range(5)` dediğimizde Python bizim için arka planda [0, 1, 2, 3, 4] sayılarını hazırlar.

```python
# İ harfi burada sırayla 0, 1, 2, 3, 4 değerlerini alacaktır.
for i in range(5):
    print(f"{i}. tur dönülüyor...")
    
print("Döngü bitti!")

# Dikkat: Tıpkı indekslemede olduğu gibi range() de saymaya 0'dan başlar
# ve yazdığımız son sayıyı (5'i) işleme DAHİL ETMEZ!
```

## 2. range() İle Başlangıç, Bitiş ve Adım Belirleme

`range()` fonksiyonunun içine 3 farklı sayı verebiliriz:
`range(başlangıç, bitiş, adım_miktarı)`

Eğer tek sayı yazarsak (Örn: range(10)), 0'dan başlar ve 1'er 1'er artar.
Eğer iki sayı yazarsak (Örn: range(2, 8)), ilk sayıdan başlar, son sayıya kadar 1'er 1'er artar.

```python
print("--- 1'den 5'e kadar (5 dahil değil) ---")
for sayi in range(1, 5):
    print("Sayı:", sayi)

print("\n--- 10'dan 50'ye kadar 10'ar 10'ar ---")
for onar in range(10, 51, 10):
    print("Adım:", onar)
    
print("\n--- Geriye doğru sayma (Eksi adım) ---")
# 5'ten başla, 0'a kadar (0 dahil değil), -1 azalarak git
for geri in range(5, 0, -1):
    print("Geri sayım:", geri)
```

## 3. while ve for Arasındaki Fark

`while` döngüsünde sayacı (adim += 1) bizim elimizle artırmamız gerekiyordu. Aksi halde sonsuz döngü oluyordu.
`for` döngüsünde ise buna hiç gerek yoktur! Python `range()` içindeki sayıları her turda otomatik olarak bir sonrakiyle değiştirir. Bu yüzden sonsuz döngü riski çok daha azdır.

```python
# while ile 3 kere "Selam" yazmak:
sayac = 0
while sayac < 3:
    print("while: Selam")
    sayac += 1
    
print("-" * 20)

# for ile 3 kere "Selam" yazmak:
for i in range(3):
    print("for: Selam")
```

## 4. String (Metin) Üzerinde Döngü Kurmak

`for` döngüsü sadece sayı üreten `range()` ile çalışmaz. Metinlerin üzerinde de harika işler çıkarır! Bir kelimeyi `for` döngüsüne verirsek, kelimenin her bir harfini tek tek alır ve işlememizi sağlar.

```python
sifre = "ROBOT"

print("Şifrenin harfleri analiz ediliyor...")
for harf in sifre:
    print("Bulunan harf:", harf)
    
# Gördüğün gibi range() kullanmadık, direkt kelimenin kendisini verdik!
```
