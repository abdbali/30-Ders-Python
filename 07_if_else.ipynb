## 1. if (Eğer) Bloğu: Sadece Şart Sağlanırsa Çalışır

`if` İngilizcede 'eğer' demektir. Kural çok basittir: if kelimesinden sonra bir şart (True/False dönecek bir durum) yazarız, sonuna iki nokta üst üste (`:`) koyarız.
Eğer şart DOĞRU (True) ise altındaki kodlar çalışır. YANLIŞ (False) ise Python o kodları tamamen atlar ve görmezden gelir.

```python
hava_yagmurlu = True

# Şartı kontrol ediyoruz. Sona İki Nokta (:) koymayı unutma!
if hava_yagmurlu == True:
    print("Dışarı çıkarken şemsiyeni almalısın.")
    
print("Bu satır if bloğunun dışında olduğu için her zaman çalışır.")
```

## 2. Girinti (Indentation) Mantığı ve IndentationError

Python, bir kodun `if` bloğuna ait olup olmadığını anlamak için süslü parantezler kullanmaz; bunun yerine **girinti (boşluk)** kullanır. 

Eğer bir kod satırı `if` satırından daha içerideyse (genellikle klavyedeki Tab tuşu veya 4 boşluk kullanılarak), Python o kodun şarta bağlı olduğunu anlar. Eğer aynı hizadaysa, şarttan bağımsızdır.

Boşlukları yanlış yaparsan Python sana `IndentationError` (Girinti Hatası) verecektir.

```python
puan = 50

if puan > 40:
    print("Sınavı geçtin!")  # Doğru girinti
    print("Tebrikler!")      # Bu da if bloğuna ait

# YANLIŞ KULLANIM (Çalıştırırsan hata verir)
# if puan > 40:
# print("Girinti yok, hata verecek!")
```

## 3. else (Değilse) Bloğu: Alternatif Yol

Peki ya şart sağlanmazsa ne olacak? Program öylece sessiz mi kalacak? 
Hayır, `else` (değilse) kullanarak "Eğer şart sağlanmazsa bunu yap" diyebiliriz.

Dikkat: `else` komutunun yanına asla bir şart yazılmaz. O, kalan tüm diğer ihtimalleri kapsar.

```python
oyuncu_yas = int(input("Lütfen yaşınızı girin: "))

if oyuncu_yas >= 18:
    print("Bu oyunu oynamaya yaşınız yetiyor, iyi oyunlar!")
else:
    # Yaş 18'den büyük değilse (yani 17 ve altındaysa) otomatik olarak burası çalışır.
    print("Üzgünüm, bu oyun yetişkinler içindir.")
```

## 4. elif (Değilse Eğer): Birden Fazla Şart

Bazen sadece iki seçeneğimiz (Siyah veya Beyaz) olmaz. Gri tonları, yani 3, 4 veya daha fazla ihtimal olabilir. 
Örneğin bir sınav notu; sadece geçti veya kaldı değil, A, B, C, D veya F olabilir.

Bu durumda `elif` (else if kısaltması) kullanırız. Python sırayla şartları kontrol eder, hangisi doğruysa onu çalıştırır ve bloğu terk eder.

```python
sicaklik = 15

if sicaklik > 30:
    print("Hava çok sıcak, tişört giy.")
elif sicaklik > 20:
    print("Hava güzel, kazak yeterli.")
elif sicaklik > 10:
    print("Hava serin, ceket giymelisin.")
else:
    print("Hava çok soğuk, montunu al!")
```

## 5. İç İçe (Nested) Karar Blokları

Bir şart sağlandığında, hemen ardından yeni bir şart daha kontrol etmek isteyebiliriz. Bunu, bir `if` bloğunun içine girintili şekilde yeni bir `if` daha yazarak yaparız.

```python
kullanici_adi = "admin"
sifre = "12345"

girilen_ad = input("Kullanıcı adınız: ")

if girilen_ad == kullanici_adi:
    # Ad doğruysa şimdi şifreyi sor
    girilen_sifre = input("Şifreniz: ")
    
    if girilen_sifre == sifre:
        print("Sisteme başarıyla giriş yaptınız.")
    else:
        print("Şifreniz yanlış!")
else:
    print("Böyle bir kullanıcı adı bulunamadı.")
```
